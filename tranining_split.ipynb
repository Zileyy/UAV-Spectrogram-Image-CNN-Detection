Imports

```python
import os
import shutil
import random
from pathlib import Path
```

Config

```python
RAW_DATA_DIR = "Data"        # your current dataset root
OUTPUT_DIR = "DataSplit"     # output folder

TRAIN_RATIO = 0.7
VAL_RATIO = 0.15
TEST_RATIO = 0.15

random.seed(42)
```

```python
def create_dir(path):
    os.makedirs(path, exist_ok=True)
```

```python
classes = [d for d in os.listdir(RAW_DATA_DIR) if os.path.isdir(os.path.join(RAW_DATA_DIR, d))]
```

```python
classes
```

Split Loop

```python
for cls in classes:
    class_path = os.path.join(RAW_DATA_DIR, cls)

    img_folder = os.path.join(class_path, "Image File")

    images = [f for f in os.listdir(img_folder) if f.lower().endswith(".png")]

    random.shuffle(images)

    n = len(images)
    train_end = int(n * TRAIN_RATIO)
    val_end = int(n * (TRAIN_RATIO + VAL_RATIO))

    train_imgs = images[:train_end]
    val_imgs   = images[train_end:val_end]
    test_imgs  = images[val_end:]

    splits = {
        "train": train_imgs,
        "val": val_imgs,
        "test": test_imgs
    }

    for split_name, split_imgs in splits.items():
        out_class_dir = os.path.join(OUTPUT_DIR, split_name, cls)
        create_dir(out_class_dir)

        for img in split_imgs:
            src = os.path.join(img_folder, img)
            dst = os.path.join(out_class_dir, img)
            shutil.copy2(src, dst)

    print(f"{cls}: {n} images → train {len(train_imgs)}, val {len(val_imgs)}, test {len(test_imgs)}")
```
